# NumPy Array Slicing

Slicing arrays
Slicing in python means taking elements from one given index to another given index.

We pass slice instead of index like this: [start:end].

We can also define the step, like this: [start:end:step].

If we don't pass start its considered 0

If we don't pass end its considered length of array in that dimension

If we don't pass step its considered 1

```python
import numpy as np
```

```python
arr = np.array([1, 2, 3, 4, 5, 6, 7])

print(arr[1:5])

"""

arr[1:5]

Here, slicing starts from index 1 (inclusive),
which means the element 2.

It ends at index 5 (exclusive),
which means the element at index 5 (6) is not included.

Therefore, Python selects elements from index 1 to index 4.

Output: [2, 3, 4, 5]
"""
```

```python
arr = np.array([1, 2, 3, 4, 5, 6, 7])

print(arr[4:])
```

```python
arr = np.array([1, 2, 3, 4, 5, 6, 7])

print(arr[:4])
```

# Negative Slicing

```python
arr = np.array([1, 2, 3, 4, 5, 6, 7])

print(arr[-3:-1])
```
